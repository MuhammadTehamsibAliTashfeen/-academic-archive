```python

import time

# Simulate the check_pin function, but with added delays to simulate timing attack vulnerability.
def check_pin(input_pin, correct_pin):
    # Simulate the time it takes to check each digit
    time_per_digit_check = 0.01  # 10 milliseconds per digit check
    for i in range(len(correct_pin)):

        time.sleep(time_per_digit_check)
        if input_pin[i] != correct_pin[i]:
            return False
    return True

# Simulate timing measurements for the correct PIN and a wrong PIN
correct_pin = "1234"
correct_pin_timings = []
wrong_pin_timings = []

start_time = time.time()
check_pin(correct_pin, correct_pin)
end_time = time.time()
correct_pin_timings.append(end_time - start_time)


wrong_pins = ["1235", "1244", "1344", "2344"]
for wrong_pin in wrong_pins:
    start_time = time.time()
    check_pin(wrong_pin, correct_pin)
    end_time = time.time()
    wrong_pin_timings.append(end_time - start_time)

correct_pin_timings, wrong_pin_timings
```
